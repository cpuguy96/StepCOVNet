## 2. (Optional) Download sample data (Colab)

Run the cell below when using this notebook in Google Colab (or when you don't have arrow training data locally). It downloads the same sample data as the training demo and sets `data_dir` to `data/train`.

```python
import os
import sys

# Project root (same as in the setup cell below)
project_root = (
    os.path.abspath(os.path.join(os.getcwd(), ".."))
    if os.path.basename(os.getcwd()) == "notebooks"
    else os.path.abspath(os.getcwd())
)
IN_COLAB = "google.colab" in sys.modules

if IN_COLAB:
    import gdown
    import zipfile

    file_id = "1dM8B30Fq-uWp-Dvi0PXGewZAnJ5T_KcP"
    url = f"https://drive.google.com/uc?id={file_id}"
    output = os.path.join(project_root, "stepcovnet_data.zip")

    print("Downloading data...")
    gdown.download(url, output, quiet=False)

    print("Unzipping...")
    data_dir_parent = os.path.join(project_root, "data")
    os.makedirs(data_dir_parent, exist_ok=True)
    with zipfile.ZipFile(output, "r") as zip_ref:
        zip_ref.extractall(data_dir_parent)

    os.remove(output)
    data_dir = os.path.join(project_root, "data", "train")
    print(f"Done! data_dir = {data_dir!r}")
else:
    # Local: use default path; override in the next cell if needed
    data_dir = os.path.join(project_root, "data", "train")
```

```python
%matplotlib inline
import os
import sys
import importlib.util

# Project root (parent of notebooks/, or current dir if run from project root)
project_root = os.path.abspath(os.path.join(os.getcwd(), "..")) if os.path.basename(os.getcwd()) == "notebooks" else os.path.abspath(os.getcwd())
if project_root not in sys.path:
    sys.path.insert(0, project_root)

# Load the visualization script as a module (no disk writes when output_dir=None, show=True)
script_path = os.path.join(project_root, "scripts", "visualize_arrow_data.py")
spec = importlib.util.spec_from_file_location("visualize_arrow_data", script_path)
viz = importlib.util.module_from_spec(spec)
spec.loader.exec_module(viz)

top_arrows = 25
```

```python
agg = viz.collect_aggregates(data_dir)
if agg is None:
    raise SystemExit(f"No audio-chart pairs found in {data_dir!r}.")

summary = viz.write_summary(agg, output_path=None)
print(summary)
```
